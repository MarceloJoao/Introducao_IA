### **1. Criando objetos:**
<font size=3>
    
* Crie a classe `Animal` com os atributos: `nome`, `especie`, `idade` e `adotado` (começando como `False`).
* Crie **três animais diferentes**.
* Mostre no console o nome e a espécie de cada um.

````python
class Animal:
    abrigo = "Abrigo Esperança"

    def __init__(self, nome, especie, idade, adotado=False):
      self.nome = nome
      self.especie = especie
      self.idade = int(idade)
      self.adotado = adotado

    def exibir_info(self):
        print(f"Nome: {self.nome}")
        print(f"Espécie: {self.especie}")
        print(f"Idade: {self.idade}")
        print(f"Adotado: {'Sim' if self.adotado else 'Não'}")

````

````python
Animal1 = Animal("Pet", "Cachorro", 3, False)
Animal2 = Animal("Brutus", "Cachorro", 4, False)
Animal3 = Animal("Bob", "Gato", 2, True)
Animal4 = Animal("Davi", "Gato", 2, True)

print(Animal1.nome, "---", Animal1.especie)
print(Animal2.nome, "---", Animal2.especie)
print(Animal3.nome, "---", Animal3.especie)
````
